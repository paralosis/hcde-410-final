# Motivation & problem 
Why this analysis, why it's interesting or useful, what you hope to learn.
I will be analyzing page views for Wikipedia in the last 6 years during two holidays, Thanksgiving and Christmas. I’m interested in seeing how the views varies between these two holidays throughout year, and how views look like before, during, and after each holiday.

# Data 
The dataset(s), what they contain, a link, the license/terms, why they fit, and any ethical considerations.
I will be using [Wikiepedia Pageviews Analysis](https://pageviews.wmcloud.org/), which contains graphs and numbers for daily page views when given a time frame and keyword for a Wikipedia page. This would help me view the page views for both holidays at the same time to view graph trends at a larger scale, and go into the daily views leading up to and after a holiday. [License](https://gitlab.wikimedia.org/repos/data-engineering/pageviews/-/blob/main/LICENSE?ref_type=heads). There could be some bias in my data analysis and interpretation, since there could be unknown factors that could've affected views throughout the year. 

# Unknowns & dependencies 
Anything outside your control that could affect finishing by term's end.
- It can be hard to determine views that are automated or human. In the FAQ for the dataset, there is a small workaround for unusually high pageviews that is caused by automated systems. But there is no definitive way for me to completely rule out human or automated views.
- For specific social media platforms, it is hard to access their user data
- Having to make an interpretation of the graphs and data without knowing for sure what affects the data.
